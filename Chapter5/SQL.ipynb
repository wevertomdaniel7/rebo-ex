```python
#biblioteca
import sqlite3
```

```python
#criar um banco de dados
conexao = sqlite3.connect('banco_de_dados.db')
```

```python
# Apontar para o banco
cursor = conexao.cursor()
```

```python
 # Criando uma tabela
 cursor.execute(
  'CREATE TABLE IF NOT EXISTS Minha_Tabela (Data TEXT, nome TEXT, idade REAL)'
 )
 #Fazer um comit
 conexao.commit()
```

```python
#inserir valores
cursor.execute('INSERT INTO Minha_Tabela VALUES("01/01/2021", "Odemir", 30)')
```
